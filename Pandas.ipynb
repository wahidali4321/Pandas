## Assignment 5 — Find Maximum and Minimum

```python
import pandas as pd

numbers = [45, 78, 23, 90, 56, 88, 34]

df = pd.Series(numbers)

ma = df.max()
mi = df.min()
su = df.sum()
av = df.mean()

print("Maximum:", ma)
print("Minimum:", mi)
print("Sum:", su)
print("Average:", av)
```

## Assignment 3 — Student Marks

```python
import pandas as pd

marks = [85, 78, 92, 69, 88]

students = ["Ali", "Ahmed", "Wahid", "Usman", "Hamza"]

df = pd.Series(marks, index=students)

print(df)

print("Wahid's marks:", df["Wahid"])
print("Hamza's marks:", df["Hamza"])
```
